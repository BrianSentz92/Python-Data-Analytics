# SAT Score Linear Regression

## Overview

This notebook demonstrates how to build a simple linear regression model using scikit-learn to predict a student's Grade Point Average (GPA) from their SAT score.

The analysis includes:

- Data exploration
- Data visualization
- Model training
- Model evaluation
- GPA prediction

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression

sns.set_theme(style="whitegrid")

Path("figures").mkdir(exist_ok=True)
```

```python
students = pd.read_csv("student_performance.csv")

students.head()
```

```python
students.info()

students.describe()
```

## Exploratory Data Analysis

Before training the model, let's examine the relationship between SAT scores and GPA.

```python
plt.figure(figsize=(8,6))

plt.scatter(
    students["SAT"],
    students["GPA"],
    alpha=.7
)

plt.title("SAT Score vs GPA")

plt.xlabel("SAT Score")

plt.ylabel("GPA")

plt.tight_layout()

plt.savefig(
    "figures/sat_vs_gpa.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
X = students[["SAT"]]

y = students["GPA"]

model = LinearRegression()

model.fit(X, y)

print(f"Intercept: {model.intercept_:.3f}")
print(f"Slope: {model.coef_[0]:.5f}")
print(f"R² Score: {model.score(X, y):.3f}")
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    students["SAT"],
    students["GPA"],
    alpha=.7,
    label="Students"
)

plt.plot(
    students["SAT"],
    model.predict(X),
    linewidth=3,
    label="Regression Line"
)

plt.title("Simple Linear Regression")

plt.xlabel("SAT Score")

plt.ylabel("GPA")

plt.legend()

plt.tight_layout()

plt.savefig(
    "figures/regression_line.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
predictions = pd.DataFrame({
    "SAT": [
        1000,
        1200,
        1400,
        1600
    ]
})

predictions["Predicted GPA"] = (
    model.predict(predictions)
)

predictions
```

```python
new_student = pd.DataFrame({
    "SAT": [1350]
})

predicted = model.predict(new_student)[0]

print(
    f"Predicted GPA for a student with a SAT score of 1350: {predicted:.2f}"
)
```
