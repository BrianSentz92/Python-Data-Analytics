# SAT Score Linear Regression

## Overview

This notebook demonstrates how to build a simple linear regression model using scikit-learn to predict a student's Grade Point Average (GPA) from their SAT score.

The analysis includes:

- Data exploration
- Data visualization
- Model training
- Model evaluation
- GPA prediction

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression

sns.set_theme(style="whitegrid")

Path("figures").mkdir(exist_ok=True)
```

```python
students = pd.read_csv("student_performance.csv")

students.head()
```

```python
students.info()

students.describe()
```

## Exploratory Data Analysis

Before training the model, let's examine the relationship between SAT scores and GPA.

```python
plt.figure(figsize=(8,6))

plt.scatter(
    students["SAT"],
    students["GPA"],
    alpha=.7
)

plt.title("SAT Score vs GPA")

plt.xlabel("SAT Score")

plt.ylabel("GPA")

plt.tight_layout()

plt.savefig(
    "figures/sat_vs_gpa.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
